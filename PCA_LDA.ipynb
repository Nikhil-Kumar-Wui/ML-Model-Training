```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# For larger dataset we can use  MNIST in Kaggle.
from sklearn.datasets import load_digits
digits= load_digits()
# this is a short dataset already in sckit learn library.
```

```python
x=digits.data
print(x.shape)
```

```python
y=digits.target
print(y.shape)
```

```python
x[:3]
```

```python
# Plotting the data for view

plt.set_cmap('binary')

for i in range(10):
  for j in range(10):
    plt.subplot(10,10,i*10+j+1)
    plt.imshow(digits.images[i*10+j])

plt.show()
plt.close()
```

#Principal Component Analysis

```python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

scaler=StandardScaler()
x_scaled=scaler.fit_transform(x)
```

```python
# Fit the PCA
pca= PCA()
pca.fit(x_scaled)
```

```python
pca.explained_variance_ratio_.round(3)
```

```python
pca.explained_variance_ratio_.cumsum().round(3)
# we can see the percentage till which variance is covered
# For example first row last PC 0.559 means 55% of the data is getting covered till first row only.
```

```python
# So for getting 90% or 95% of information/variance we need to go till 0.9 or 0.951
```

```python
# Fitting by defining the numebr of components wanted
pca= PCA(n_components=2)
X_embedded_2comp=pca.fit_transform(x_scaled)
```

```python
X_embedded_2comp
```

```python
pca= PCA(n_components=0.9)
X_embedded_90var=pca.fit_transform(x_scaled)
```

```python
X_embedded_90var
```

```python
X_embedded_90var.shape
```

```python
# Plot all the digits using 2 components from PCA
plt.set_cmap('tab10')

fig, ax=plt.subplots()
scatter= ax.scatter(X_embedded_2comp[:,0],X_embedded_2comp[:,1],c=digits.target)

legend=ax.legend(*scatter.legend_elements(), title='Digit')
ax.add_artist(legend)

plt.show()
plt.close()
```

#LDA

```python
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis as LDA
```

```python
# Fit the data
lda=LDA()
X_embedded_lda=lda.fit_transform(x_scaled,y)
```

```python
lda.explained_variance_ratio_.round(3)
```

```python
lda.explained_variance_ratio_.cumsum().round(3)
```

```python
# Fit the data
lda=LDA(n_components=2)
X_embedded_lda_2comp=lda.fit_transform(x_scaled,y)
```

```python
X_embedded_lda_2comp.shape
```

```python
# Plot all the digits using 2 components from PCA
plt.set_cmap('tab10')

fig, ax=plt.subplots()
scatter= ax.scatter(X_embedded_lda_2comp[:,0],X_embedded_lda_2comp[:,1],c=digits.target)

legend=ax.legend(*scatter.legend_elements(), title='Digit')
ax.add_artist(legend)

plt.show()
plt.close()
```

```python
"""
Whatever your problem wrt ML->
You can use x_embedded_2Comp or x_embedded_lda_2comp or x_embedded_lda or etc, as features for your model model.

Instead of x which is high dimension you can now use lower dimension as independent features in your model
"""
```
