```python
for col in df.columns:
  if df[col].dtype!="object":
    plt.boxplot(df[col])
    plt.title(col)
    plt.show()
```

```python
out_cols=['wheelbase','carlength','carwidth','enginesize','stroke','compressionratio','horsepower','peakrpm','citympg','highwaympg','price']
```

```python
for col in out_cols:
  q1=df[col].quantile(0.25)
  q3=df[col].quantile(0.75)
  iqr=q3-q1
  upper_limit=q3+(1.5*iqr)
  lower_limit=q1-(1.5*iqr)
  df=df[(df[col]<=upper_limit) & (df[col]>=lower_limit)]
```

```python
df
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le=LabelEncoder()
```

```python
object_cols=['name','fueltypes', 'aspiration','doornumbers','carbody','drivewheels','enginelocation','enginetype','cylindernumber','fuelsystem' ]
for cols in object_cols:
  df[cols]=le.fit_transform(df[cols])
```

```python
imp_cols=['drivewheels','wheelbase','carlength', 'carwidth','curbweight','enginesize', 'fuelsystem', 'boreratio','horsepower']
```

```python
#Independent and Dependent

# Independent Variable
x=df[imp_cols]
# Dependent Variable
y=df['price']
x
```

```python
y
```

```python
# Splitting the Data into Training and Testing

from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=52)
```

```python
x_train
```

```python
x_test
```

```python
y_test
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model=LinearRegression()
```

```python
# Training Machine
model.fit(x_train,y_train)
```

```python
# Testing the Machine
y_pred=model.predict(x_test)
y_pred
```

```python
y_test
```

```python
from sklearn.metrics import r2_score,mean_squared_error
```

```python
mean_squared_error(y_test,y_pred)
```

```python
r2_score(y_test,y_pred)
```

```python
np.sqrt(mean_squared_error(y_test,y_pred))
```
